## 1.6. Calculate time offset in days - part 2
Great work! Now, we have six different data sets with year, month and day values for Invoice and Cohort dates - `invoice_year`, `cohort_year`, `invoice_month`, `cohort_month`, `invoice_day`, and `cohort_day`.

In this exercise you will calculate the difference between the Invoice and Cohort dates in years, months and days separately and then calculate the total days difference between the two. This will be your days offset which we will use in the next exercise to visualize the customer count. The `online` data has been loaded, you can print its header to the console by calling `online.head()`.

### Instructions
- Find the difference in years between the moment of the invoice and the moment the cohort was created.
- Find the difference in months between the moment of the invoice and the moment the cohort was created.
- Find the difference in days between the moment of the invoice and the moment the cohort was created.
- Calculate the number of days for the `CohortIndex` (assume 365 days in a year, and 30 days in a month).

```python
# Calculate difference in years
years_diff = invoice_year - cohort_year

# Calculate difference in months
months_diff = invoice_month - cohort_month

# Calculate difference in days
days_diff = invoice_day - cohort_day

# Extract the difference in days from all previous values
online['CohortIndex'] = years_diff * 365 + months_diff * 30 + days_diff + 1
print(online.head())
```

> Perfect! You are now fully equipped to apply RFM segmentation to any dataset!

---

## Course interval:

## I. Cohort Analysis: Longitudinal Dynamics & Lifecycle-Time Normalization

### 1. First-Principles Abstraction

In biostatistics and reliability engineering, survival analysis investigates the time elapsed before an entity experiences a defined terminal event (e.g., death, failure, or system breakdown). **Cohort analysis** applies a related longitudinal tracking framework to customer behavior by grouping entities according to a shared initial event and observing their behavioral persistence over subsequent periods.

Its fundamental objective is to distinguish lifecycle progression from absolute calendar time. Rather than comparing customers solely by when an observation occurred, cohort analysis realigns temporal observations relative to a common starting event: **Lifecycle Time ($t_{\text{offset}}$)**.

This transformation enables comparisons between cohorts at equivalent lifecycle stages, reducing the distortion introduced by differences in acquisition dates. However, cohort alignment does not eliminate calendar-time effects such as seasonality, macroeconomic conditions, marketing campaigns, or product changes. These factors may still influence observed retention patterns.

### 2. The Architectural Mechanics

* **Temporal Anchoring (State 0):** Every customer is assigned to a discrete cohort based on a predefined entry event, typically their first transaction. The acquisition timestamp establishes the temporal origin for that customer, while the acquisition period defines the cohort identifier ($C_i$).

* **Lifecycle-Time Transformation:** Absolute transaction timestamps are converted into relative lifecycle offsets using integer arithmetic:

$$
\Delta t =
(\text{Year}_{\text{invoice}}-\text{Year}_{\text{acquisition}})\times12
+
(\text{Month}_{\text{invoice}}-\text{Month}_{\text{acquisition}})
+1
$$

The acquisition month is conventionally designated as Lifecycle Month 1, followed by subsequent calendar months.

* **Retention Matrix Construction:** Grouping observations by $(C_i,\Delta t)$ and counting unique active customers produces a cohort-by-lifecycle-period matrix. The resulting matrix typically exhibits a triangular observation structure because recently acquired cohorts have fewer observable lifecycle periods.

The empirical period-specific retention rate is calculated as:

$$
R(C_i,\Delta t)=
\frac{N_{\text{active}}(C_i,\Delta t)}
{N_{\text{initial}}(C_i)}
$$

Where:

- $N_{\text{active}}(C_i,\Delta t)$ represents the number of unique customers from cohort $C_i$ who performed at least one qualifying activity during lifecycle period $\Delta t$.
- $N_{\text{initial}}(C_i)$ represents the initial population of customers assigned to cohort $C_i$.

This formulation measures **period-specific retention**, not necessarily continuous survival. A customer may become inactive during one period and return in a later period.

For retention metrics that require continuous survival, the event definition and censoring mechanism must be explicitly specified, potentially using formal survival-analysis estimators.



## II. RFM Analysis: Multidimensional Behavioral Representation & Ordinal Discretization

### 1. First-Principles Abstraction

While cohort analysis examines population-level behavioral evolution over time, **RFM (Recency, Frequency, Monetary) analysis** constructs a cross-sectional behavioral representation of individual customers using historical transactional records within a predefined observation window (e.g., $T=12$ months).

It transforms transactional histories into a three-dimensional behavioral feature space:

$$
\vec{v}_k=(R_k,F_k,M_k)
$$

Where:

* **Recency ($R$):** The elapsed time between the evaluation timestamp ($T_{\text{snapshot}}$) and the customer's most recent qualifying transaction. It measures purchase inactivity duration, where a lower value generally indicates more recent engagement.

$$
R_k=T_{\text{snapshot}}-T_{\text{last purchase},k}
$$

* **Frequency ($F$):** The total number of qualifying transactions performed by customer $k$ during the observation window. It represents transactional intensity, where higher values generally indicate more frequent purchasing behavior.

$$
F_k=\sum_{j=1}^{n_k}1
$$

* **Monetary ($M$):** The cumulative monetary value of qualifying transactions generated by customer $k$ during the observation window. It represents historical monetary contribution, where higher values indicate greater transactional spending.

$$
M_k=\sum_{j=1}^{n_k}\text{TransactionValue}_{k,j}
$$

The resulting feature vector provides a compact representation of customer behavior at a specific evaluation epoch. Although the representation is cross-sectional, its underlying features are derived from longitudinal transaction histories.

### 2. Discretization & Non-Parametric Quantile Binning

Transactional data frequently exhibits skewed distributions, heavy tails, and extreme observations. Applying uniform numerical thresholds may therefore produce highly imbalanced customer groups.

To transform continuous behavioral measurements into ordinal categorical representations, we apply **empirical quantile-based discretization**, commonly implemented using the ECDF or Pandas `qcut`.

For example, dividing each feature into five quantile groups produces ordinal scores ranging from 1 to 5.

* **Directional Alignment:** Because Recency has an inverse relationship with customer activity, while Frequency and Monetary generally have positive relationships with customer value, the ordinal mapping must account for directional differences:

  - Recency: Score 5 = lowest recency values (most recently active customers).
  - Frequency: Score 5 = highest transaction counts.
  - Monetary: Score 5 = highest cumulative transaction values.

* **Ordinal Representation:** Each customer's continuous behavioral measurements are transformed into a discrete three-dimensional ordinal profile.

$$
\text{RFM Profile}_k=(R_{\text{rank}},F_{\text{rank}},M_{\text{rank}})
$$

For example:

- `555`: Recently active, highly frequent, and high-spending customer.
- `444`: Relatively recent, frequent, and high-spending customer.
- `111`: Less recently active, infrequent, and low-spending customer.

* **Composite Score Aggregation:** An optional scalar index can be constructed by summing the three ordinal dimensions:

$$
\text{RFM Score}_k=R_{\text{rank}}+F_{\text{rank}}+M_{\text{rank}}
$$

This produces a heuristic customer-value index rather than a statistically calibrated estimate of Customer Lifetime Value (CLV).

Importantly, customers with identical composite scores may exhibit substantially different behavioral profiles. Therefore, retaining the original three-dimensional RFM representation is preferable when performing detailed customer segmentation.

* **Customer Segmentation:** The resulting ordinal profiles can be mapped into predefined business categories such as Champions, Loyal Customers, Potential Loyalists, At Risk, and Hibernating Customers. These classifications are business-defined segmentation rules rather than inherently discovered statistical clusters.



## III. Structural Paradigm Synthesis

| Theoretical Dimension | Cohort Analysis | RFM Analysis |
|---|---|---|
| **System Viewpoint** | Longitudinal / Cohort-Based | Cross-Sectional / Customer-Level |
| **Data Representation** | Cohort × Lifecycle-Period Matrix | Three-Dimensional Behavioral Feature Vector |
| **Mathematical Domain** | Retention Rates & Longitudinal Analysis | Ordinal Statistics & Quantile Discretization |
| **Primary Objective** | Measure lifecycle retention and cohort evolution | Segment customers by transactional behavior and monetary contribution |
| **Key Transformation** | Calendar Time → Lifecycle-Relative Time | Continuous Features → Ordinal Categories |
| **Analytical Granularity** | Population / Cohort Level | Individual Customer Level |
| **Temporal Reference** | Acquisition Event | Evaluation Snapshot |
| **Analytical Output** | Retention Curves / Cohort Heatmaps | RFM Profiles / Customer Segments |



## IV. Integrated Analytical Perspective: Temporal × Behavioral Intelligence

Although Cohort Analysis and RFM Analysis operate on different analytical dimensions, they are complementary rather than competing methodologies.

Cohort Analysis provides a longitudinal perspective on population-level retention dynamics, whereas RFM provides a cross-sectional representation of individual customer behavior and monetary contribution.

Combining both approaches enables **lifecycle-aware customer intelligence**.

For example, RFM segmentation can be performed independently within each acquisition cohort, allowing analysts to examine how the composition of customer segments evolves as cohorts mature.

This integration supports questions such as:

- Which acquisition cohorts generate the highest proportion of high-value customers?
- How does the proportion of Champions or Loyal Customers change across lifecycle periods?
- Do customers classified as At Risk exhibit different retention patterns across acquisition cohorts?
- Are newer cohorts demonstrating stronger transactional engagement than older cohorts at equivalent lifecycle stages?

Ultimately, the two methodologies provide complementary perspectives:

**Cohort Analysis answers how customer populations evolve over time, while RFM Analysis answers what individual customer behavioral profiles look like at a given evaluation point.**

Together, they establish a foundation for understanding both customer lifecycle dynamics and customer-level behavioral value.

---

# 3. Data Preprocessing for Clustering

Once you created some segments, you want to make predictions. However, you first need to master practical data preparation methods to ensure your k-means clustering algorithm will uncover well-separated, sensible segments.

## 3.1. Data preprocessing
In this lesson, you'll explore key data pre-processing steps for effective k-means clustering on RFM numbers to identify customer segments:

- Understand the importance of symmetrical distributions, equal average values, and variance scaling for variables.
- Learn to identify skewed distributions visually and apply logarithmic transformations.
- Use the "describe" function to examine key statistics of your data.

Here's a quick code snippet to get you started:
```python
import pandas as pd

# Load your data
data = pd.read_csv('your_data.csv')

# Check statistics
print(data.describe())

# Apply log transformation
data['log_variable'] = data['variable'].apply(lambda x: np.log(x) if x > 0 else x)
```

## III. Data Preprocessing for Clustering: Statistical Transformation & Distance-Space Normalization

### 1. First-Principles Abstraction

In unsupervised machine learning, clustering algorithms identify structural similarities among observations by measuring distances within a feature space. **Data preprocessing for clustering** transforms raw numerical features into a representation where these distances are statistically meaningful and comparable.

K-means clustering relies on Euclidean distance and minimizes the within-cluster sum of squared distances between observations and their respective centroids. Consequently, the numerical distribution, variance, and scale of each feature directly influence the resulting cluster geometry.

Transactional customer data frequently exhibits highly skewed distributions, extreme values, and substantial differences in measurement scales. For example, Recency may be measured in days, Frequency in transaction counts, and Monetary in currency units. Using these raw measurements directly can cause high-magnitude variables to dominate the distance calculation.

The preprocessing pipeline therefore addresses three practical concerns:

- **Distributional Transformation:** Reducing excessive skewness and the influence of extreme values.
- **Feature Standardization:** Bringing numerical variables onto comparable scales.
- **Representation Preservation:** Maintaining the original measurements separately for subsequent cluster profiling and business interpretation.

Importantly, symmetry and equal feature means are not strict mathematical requirements of K-means. Rather, K-means works most naturally when the underlying cluster structure is reasonably compatible with its Euclidean, centroid-based geometry. Standardization is a practical strategy for reducing unintended scale dominance, not a guarantee of meaningful clusters.

---

### 2. The Architectural Mechanics

#### A. Distribution Diagnostics & Skewness Identification

Before transformation, each feature must be examined to understand its distribution and statistical properties.

Common diagnostic techniques include:

- Descriptive statistics using `describe()` to inspect the mean, standard deviation, minimum, maximum, and quartiles.
- Distribution plots such as histograms and density curves to identify skewness.
- Skewness statistics to quantify distributional asymmetry.

Customer transactional variables frequently exhibit positive (right) skewness, where most observations concentrate around relatively small values while a smaller number of customers generate exceptionally large values.

For example:

- Recency may exhibit a long right tail due to customers who have not purchased for extended periods.
- Frequency may exhibit substantial right skewness because most customers transact relatively infrequently while a smaller population purchases repeatedly.
- Monetary may exhibit extreme right skewness due to a concentration of high-spending customers.

---

#### B. Distributional Transformation: Logarithmic Normalization

A logarithmic transformation is commonly applied to positively skewed transactional variables because it compresses large values more aggressively than small values.

For strictly positive observations:

$$
X'=\log(X)
$$

For non-negative observations that include zeros, an alternative is:

$$
X'=\log(1+X)
$$

The transformation reduces the influence of large numerical magnitudes and can produce a more balanced distribution.

However, logarithmic transformation does not guarantee normality or perfect symmetry. Its effectiveness depends on the original distribution.

For variables containing negative observations, alternative approaches include:

- Yeo-Johnson transformation, which accommodates positive, zero, and negative values.
- Cube-root transformation, which can preserve negative values.
- Carefully defined constant shifting before logarithmic transformation.

Constant shifting must be applied cautiously because it changes the transformed distances and makes the result dependent on the chosen offset. Negative transactional values should also be investigated semantically, particularly when they represent refunds, chargebacks, or accounting adjustments.

---

#### C. Feature Centering & Variance Scaling

Once the distributions have been examined and transformed where appropriate, numerical variables must be brought onto comparable scales.

The central concern is not that K-means requires identical feature means. Rather, features with substantially different variances can contribute disproportionately to Euclidean distance.

The standardization process involves two operations:

**1. Centering**

Subtract the feature mean from every observation:

$$
X_{\text{centered}}=X-\mu_X
$$

This produces a transformed variable with a mean approximately equal to zero.

**2. Standard Deviation Scaling**

Divide the centered observations by their standard deviation:

$$
Z=\frac{X-\mu_X}{\sigma_X}
$$

This produces standardized variables with approximately zero mean and unit variance.

The combined procedure is commonly called **Z-score Standardization**.

In Python, this can be implemented using `StandardScaler()` from `sklearn.preprocessing`.

The resulting standardized representation allows features to contribute on a more comparable numerical scale when Euclidean distances are calculated.

An important mathematical distinction is that centering alone does not change pairwise Euclidean distances:

$$
\|(x_i-\mu)-(x_j-\mu)\|_2
=
\|x_i-x_j\|_2
$$

Therefore, scaling, rather than centering itself, is what changes the relative influence of features in K-means.

Standardization also does not guarantee that all features receive equal business importance. Highly correlated variables can still collectively give greater influence to the same underlying behavioral dimension.

---

#### D. Sequential Preprocessing Pipeline

The order of transformations matters because different operations have different numerical requirements.

The recommended pipeline is:

**1. Distributional Transformation**

Apply logarithmic or other appropriate power transformations to reduce skewness before standardization.

**2. Centering and Scaling**

Apply Z-score standardization to the transformed variables.

**3. Representation Preservation**

Store the transformed and standardized data separately from the original RFM dataset.

The complete transformation can be expressed as:

$$
X_{\text{raw}}
\longrightarrow
X_{\text{transformed}}
\longrightarrow
Z_{\text{standardized}}
$$

The three representations serve different purposes:

| Representation | Purpose |
|---|---|
| Raw RFM | Business interpretation and cluster profiling |
| Transformed RFM | Reduced distributional skewness |
| Standardized RFM | Distance computation and K-means clustering |

Maintaining separate representations is essential because standardized values no longer retain their original business units.

For example, a standardized Monetary value of 1.5 does not mean the customer spent 1.5 units of currency. It indicates that the transformed monetary measurement is 1.5 standard deviations above its reference mean.

For production-oriented implementations, transformation and scaling parameters should also be retained. New customers must be transformed using the same reference parameters rather than independently fitting a new scaler.

## IV. Customer Segmentation with K-means: Unsupervised Partitioning & Segment Intelligence

### 1. First-Principles Abstraction

In unsupervised learning, clustering identifies groups of observations that exhibit similar characteristics without requiring predefined target labels.

**K-means clustering** is a centroid-based partitioning algorithm that divides a dataset into a predefined number of clusters ($K$), assigning each observation to the nearest cluster centroid according to Euclidean distance.

Its fundamental objective is to minimize the **Within-Cluster Sum of Squares (WCSS)**, also commonly referred to as Sum of Squared Errors (SSE) or inertia.

Mathematically:

$$
J(K)=
\sum_{k=1}^{K}
\sum_{i\in C_k}
\|x_i-\mu_k\|_2^2
$$

Where:

- $K$ represents the number of clusters.
- $C_k$ represents the set of observations assigned to cluster $k$.
- $x_i$ represents an individual customer feature vector.
- $\mu_k$ represents the centroid of cluster $k$.
- $\|\cdot\|_2$ represents Euclidean distance.

The algorithm iteratively assigns customers to their nearest centroid and recomputes each centroid as the arithmetic mean of its assigned observations.

This process continues until the assignments stabilize or a stopping criterion is reached.

The resulting cluster labels represent discovered behavioral groupings rather than predefined customer categories or probabilities of future outcomes.

---

### 2. The Architectural Mechanics

#### A. Determining the Number of Clusters ($K$)

K-means requires the number of clusters to be specified before model fitting. Selecting an appropriate value is therefore an important model-selection decision.

Several techniques can support this decision.

**1. Elbow Criterion**

The elbow method evaluates the total within-cluster sum of squared distances across multiple candidate values of $K$.

$$
WCSS(K)=
\sum_{k=1}^{K}
\sum_{i\in C_k}
\|x_i-\mu_k\|_2^2
$$

As the number of clusters increases, WCSS decreases because observations have more centroids available to represent their positions.

However, the reduction becomes progressively smaller, producing diminishing returns.

The elbow point represents a region where adding more clusters results in relatively modest improvements in within-cluster compactness.

It provides a practical estimate of the number of clusters, but does not establish a mathematically unique optimal solution.

**2. Silhouette Coefficient**

The silhouette coefficient measures how well an observation fits within its assigned cluster relative to its nearest alternative cluster.

For observation $i$:

$$
s(i)=\frac{b(i)-a(i)}{\max(a(i),b(i))}
$$

Where:

- $a(i)$ is the average distance between observation $i$ and other observations in its own cluster.
- $b(i)$ is the minimum average distance between observation $i$ and observations in another cluster.

Interpretation:

- Values approaching +1 indicate well-separated cluster assignments.
- Values near 0 indicate overlapping or ambiguous boundaries.
- Negative values suggest that an observation may be closer to another cluster.

Although useful for assessing cluster cohesion and separation, silhouette scores should not replace business interpretation.

**3. Experimental and Business-Driven Selection**

After identifying candidate values through quantitative methods, several segmentation solutions should be constructed and compared.

For example, if the elbow method suggests two clusters, it is still useful to examine three- and potentially four-cluster alternatives.

The most appropriate solution should balance:

- Within-cluster cohesion.
- Between-cluster separation.
- Segment stability.
- Interpretability.
- Business relevance and actionability.

The largest number of clusters or the lowest WCSS does not automatically produce the most valuable segmentation.

---

#### B. Model Initialization & Cluster Assignment

Once the number of clusters has been selected, the K-means algorithm can be fitted using the standardized customer feature matrix.

The main implementation stages are:

1. Initialize $K$ cluster centroids.
2. Calculate distances between observations and centroids.
3. Assign each observation to its nearest centroid.
4. Recalculate centroids using the mean of assigned observations.
5. Repeat the assignment and centroid-update procedures until convergence.

The resulting output contains a cluster label for every customer.

In scikit-learn, the `KMeans` estimator provides the implementation. Specifying `random_state` and multiple initializations using `n_init` helps improve reproducibility and reduces dependence on an unfavorable initialization.

Cluster labels such as 0, 1, and 2 are arbitrary identifiers. They do not inherently imply customer quality, priority, or value.

---

#### C. Cluster Profiling Using Original Measurements

The next step is to understand what differentiates the discovered customer groups.

Although clustering is performed using standardized features, the resulting labels should be joined back to the original dataset for business profiling.

The primary profiling procedure consists of:

1. Attaching cluster labels to the original customer-level RFM dataset.
2. Grouping customers by cluster.
3. Calculating average Recency, Frequency, and Monetary values for each cluster.
4. Counting the number of customers assigned to each segment.
5. Comparing the resulting segment profiles.

The cluster-level descriptive statistics reveal the behavioral characteristics of each group.

For example:

- A cluster with low Recency, high Frequency, and high Monetary represents highly engaged, high-spending customers.
- A cluster with high Recency, low Frequency, and low Monetary represents less active, lower-value customers.

Means provide an initial summary, but medians, quartiles, and distribution plots can provide additional insight when the underlying data remains highly skewed.

Cluster size is also important because a mathematically distinct segment may have limited commercial relevance if it contains very few customers.

---

#### D. Segment Visualization: Snake Plots

A **Snake Plot** is a comparative visualization technique used to examine the behavioral profiles of different customer segments across multiple features.

It represents the average value of each feature for every cluster using a line chart.

Because Recency, Frequency, and Monetary have different original measurement scales, their normalized representations should be used for meaningful visual comparison.

The construction process involves:

1. Reconstructing a DataFrame from the standardized feature matrix.
2. Attaching the corresponding cluster labels.
3. Aggregating the standardized feature values by cluster.
4. Reshaping the dataset from wide format into long format using `melt()`.
5. Plotting customer segments against their average feature values.

The resulting visualization makes it easier to identify relative behavioral differences between clusters.

For example, one segment may exhibit consistently above-average Frequency and Monetary values, while another may exhibit substantially higher Recency.

The Snake Plot is particularly useful for identifying distinctive segment profiles that may not be immediately apparent from summary tables.

---

#### E. Relative Importance of Segment Attributes

Another profiling technique evaluates how strongly each feature differs within a particular cluster relative to the overall customer population.

For a positive-valued feature with a non-zero population mean, the relative deviation is calculated as:

$$
RI_{k,j}=
\frac{\bar{X}_{k,j}}
{\bar{X}_{\text{population},j}}-1
$$

Where:

- $\bar{X}_{k,j}$ represents the average value of feature $j$ within cluster $k$.
- $\bar{X}_{\text{population},j}$ represents the overall population average of that feature.

Interpretation:

- Positive values indicate that the cluster's average is above the population average.
- Negative values indicate that the cluster's average is below the population average.
- Values close to zero indicate that the feature is relatively similar to the overall population.

For example:

$$
RI_{k,\text{Monetary}}=0.40
$$

Indicates that the cluster's average Monetary value is 40% higher than the overall population average.

These values can be visualized using a heatmap, making it easier to identify distinctive behavioral characteristics across segments.

This measure represents relative deviation from the population average, not statistical significance or causal feature importance. It is best calculated using original positive-valued customer metrics rather than standardized features with approximately zero means.

---

#### F. Extended Feature Space: Incorporating Customer Tenure

The final segmentation exercise introduces an additional variable: **Customer Tenure**.

Tenure measures the elapsed time between a customer's first transaction and the evaluation timestamp.

$$
\text{Tenure}_k=
T_{\text{snapshot}}-
T_{\text{first purchase},k}
$$

Although Recency and Tenure are both time-related measurements, they describe different aspects of customer behavior:

- Recency measures how long ago the customer last purchased.
- Tenure measures how long ago the customer first entered the transactional system.

Combining Tenure with RFM expands the feature vector:

$$
\vec{v}_k=(R_k,F_k,M_k,T_k)
$$

This provides an additional dimension for differentiating customers based on lifecycle maturity.

For example, two customers may have identical Recency, Frequency, and Monetary values but substantially different Tenure. One may be a relatively new customer, whereas the other may have maintained a relationship with the company for several years.

However, adding Tenure also changes the geometry of the clustering problem. Its correlation with existing features, particularly Recency and cohort membership, should be examined to avoid unintentionally overweighting similar behavioral information.

The expanded feature set must undergo the same diagnostic, transformation, and standardization procedures before clustering.

## VI. Integrated Analytical Architecture: From Customer Behavior to Actionable Segmentation

The four chapters form a connected analytical workflow, with each methodology addressing a different layer of the customer analytics problem.

| Chapter | Analytical Framework | Core Question | Principal Output |
|---|---|---|---|
| **I. Cohort Analysis** | Longitudinal Customer Analysis | How do customer populations evolve over time? | Retention Matrix / Cohort Heatmap |
| **II. RFM Analysis** | Behavioral Feature Engineering | What are the transactional characteristics of individual customers? | RFM Profiles / Customer Segments |

### 1. End-to-End Workflow

The complete analytical architecture can be expressed as:

**Transaction Records → Customer-Level Aggregation → Behavioral Analysis → Feature Transformation → Clustering → Segment Profiling → Business Interpretation**

The workflow consists of the following stages:

**Stage 1 — Data Foundation & Temporal Structuring**

Collect transactional records and establish customer identities, transaction timestamps, monetary values, and a consistent evaluation period.

Use Cohort Analysis to establish acquisition-based populations and examine customer retention across equivalent lifecycle periods.

---

**Stage 2 — Behavioral Feature Construction**

Construct the RFM dataset at the customer level using a predefined observation window and evaluation timestamp.

Optionally extend the behavioral representation with additional features such as Customer Tenure.

---

**Stage 3 — Distributional Diagnostics & Preprocessing**

Examine feature distributions, apply appropriate transformations to skewed variables, and standardize the resulting feature matrix.

Preserve the original customer-level measurements for subsequent interpretation.

---

**Stage 4 — Model Selection & Clustering**

Evaluate candidate values of $K$ using the elbow criterion and, where appropriate, silhouette analysis.

Fit K-means using the selected feature matrix and obtain cluster assignments for each customer.

---

**Stage 5 — Segment Profiling & Evaluation**

Compare cluster-level RFM and Tenure statistics using:

- Mean and median descriptive statistics.
- Cluster population counts.
- Snake plots.
- Relative importance analysis.
- Heatmaps.

Evaluate whether the resulting segments exhibit meaningful behavioral differences and sufficient business relevance.

---

**Stage 6 — Business Interpretation & Actionability**

Translate the mathematical cluster profiles into descriptive customer segments and identify appropriate commercial strategies.

Examples include retention campaigns for at-risk customers, loyalty programs for frequent purchasers, and targeted engagement for high-value customers.

---

### 2. Critical Integration: The Different Roles of Each Methodology

The four analytical approaches should not be interpreted as interchangeable techniques.

Cohort Analysis introduces a temporal population perspective. RFM transforms transaction histories into customer-level behavioral measurements. Data preprocessing prepares those measurements for distance-based comparisons. K-means then partitions customers according to similarities in the resulting feature space.

An important distinction is that **quantile-based RFM scoring and K-means feature preparation are two different representations of customer behavior**.

RFM quantization converts continuous metrics into ordinal scores for rule-based customer segmentation. K-means generally benefits from using the underlying continuous RFM measurements after appropriate distributional transformation and scaling.

Consequently, ordinal RFM scores should not automatically replace raw continuous RFM measurements in K-means. Doing so changes the geometry of the feature space and can discard information about the magnitude of behavioral differences.

The two representations serve different analytical purposes and can coexist within the same customer analytics architecture.

---

### 3. Final Conceptual Synthesis

The integrated workflow establishes four complementary analytical perspectives:

- **Cohort Analysis = Temporal Intelligence**: Understand how customer populations evolve relative to their acquisition periods and lifecycle stages.

- **RFM Analysis = Behavioral Intelligence**: Represent individual customer behavior using purchase recency, transaction frequency, and monetary contribution.

- **Data Preprocessing = Numerical Representation Intelligence**: Transform and standardize customer features to make distance-based comparisons more appropriate and interpretable.

- **K-means = Structural Intelligence**: Discover customer groupings based on similarities in their multidimensional behavioral characteristics.

Together, these methodologies provide a systematic foundation for customer segmentation:

**From understanding when customers enter and how they persist, to understanding what they do, how they differ, and which behavioral groups can be addressed through distinct business strategies.**

## Additional technical observations

There are five points I would particularly retain when moving from the course material to an actual Data Science implementation.

| Topic               | Important consideration                                                                                                                                           |
| ------------------- | ----------------------------------------------------------------------------------------------------------------------------------------------------------------- |
| K-means assumptions | It does not require equal means or perfectly symmetrical feature distributions. It is sensitive to feature scale, outliers, and non-spherical cluster structures. |
| Log transformation  | Use `log1p()` when zero values are valid. Do not blindly shift negative financial values without understanding their meaning.                                     |
| StandardScaler      | Centering alone does not change pairwise Euclidean distances; variance scaling changes the relative feature contributions.                                        |
| Cluster selection   | The elbow is a heuristic, not proof of an optimal K. Business interpretability and segment stability matter.                                                      |
| Cluster profiling   | Always return to original measurements for business interpretation. Standardized values are not monetary or transactional quantities.                             |
